```python
import ast

def parse_to_list(data_line):
  if data_line.startswith('"') and data_line.endswith('"'):
      data_line = data_line[1:-1]

  if data_line.startswith("'") and data_line.endswith("'"):
      data_line = data_line[1:-1]

  data_line = data_line.replace("\\'", "'").replace('\\"', '"')

  try:
      parsed_data = ast.literal_eval(data_line)
      return parsed_data
  except Exception as e:
      print("Error al parsear:", e)

def read_RankDivergence(file_path):


    context_data = {}

    with open(file_path, 'r') as f:
        lines = f.readlines()

    for i in range(0, len(lines), 2):
        meta = lines[i].strip().split(",")
        try:
            divergence = float(meta[0].strip())
        except ValueError:
            print(f"Error al leer divergence en línea {i+1}: {lines[i]}")
            continue

        context = meta[1].strip()
        data_line = lines[i + 1].strip()

        try:
            parsed_data = parse_to_list(data_line)
            context_data[context] = {
                'divergence': divergence,
                'data': parsed_data
            }
        except Exception as e:
            print(f"Error al convertir la línea {i+2} a lista de diccionarios:\n{data_line}\nError: {e}")

    return context_data
```

```python
import csv
import ast
from collections import defaultdict

def read_RankPerplexity_v2(filepath):
    results = defaultdict(dict)

    with open(filepath, newline='', encoding='utf-8') as csvfile:
        reader = csv.reader(csvfile)
        for row in reader:
            if len(row) != 3:
                continue  # skip malformed rows
            context, model_name, data_str = row
            try:
                parsed_data = ast.literal_eval(data_str)  # safely parse stringified list of dicts
            except Exception as e:
                print(f"Error parsing row for context '{context}', model '{model_name}': {e}")
                continue

            # Optional: compute divergence if needed
            divergence = sum(item["perplexity"] for item in parsed_data)

            results[model_name][context] = {
                'data': parsed_data,
                'divergence': divergence
            }

    return results
```

```python
import matplotlib.pyplot as plt
import plotly.graph_objects as go

context_data = read_RankDivergence("experiment4/DPSGD4_Rank+DivergenceExperiment")

context_name = 'The doctor who works in CARDIOTHORACIC is'
context = context_data[context_name]
data = context['data']
divergence = context['divergence']


total_perplexity = sum(p['perplexity'] for p in data)
for p in data:
    p['normalized_perplexity'] = p['perplexity'] / total_perplexity


data = sorted(data, key=lambda x: x['normalized_perplexity'], reverse=True)


top_n = 20
top_data = data[:top_n]
others_data = data[top_n:]


others_sum = sum(p['normalized_perplexity'] for p in others_data)


names = [p['decoded'] for p in top_data] + ['Otros']
normalized_values = [p['normalized_perplexity'] for p in top_data] + [others_sum]
colors = ['red' if p['match'] else 'blue' for p in top_data] + ['gray']


fig = go.Figure(data=[go.Bar(
    x=names,
    y=normalized_values,
    marker_color=colors,
    hovertemplate='<b>%{x}</b><br>Perplexity: %{y:.6f}<extra></extra>'
)])

fig.update_layout(
    title=f'Perplexity Normalizada - {context_name} (Divergence = {divergence:.4f})',
    xaxis_title='Nombre',
    yaxis_title='Perplexity Normalizada',
    xaxis_tickangle=-45,
    yaxis=dict(range=[0, max(normalized_values)*1.2]),
    height=600,
    margin=dict(b=150)
)
fig.show()
```

```python
import pandas as pd
import ast
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv('experiment4/Laura_Rank+DivergenceExperiment.txt')

def extract_perplexity(result_str):
    try:
        result_list = ast.literal_eval(result_str)
        if isinstance(result_list, list) and len(result_list) > 0:
            return result_list[0].get('perplexity', None)
    except Exception as e:
        print(f"Error parsing result: {e}")
        return None

df['perplexity'] = df['result'].apply(extract_perplexity)


df['model'] = df['model'].replace({
    'SGD with Laura': 'SGD with Laura',
    'SGD without Laura': 'SGD without Laura',
    'DPSGD with Laura': 'DPSGD with Laura',
    'DPSGD without Laura': 'DPSGD without Laura'
})


custom_palette = {
    'SGD_withLaura': '#d73027',     
    'SGD_withoutLaura': '#fdae61',     
    'DPSGD_withLaura': '#4575b4',     
    'DPSGD_withoutLaura': '#a6bddb'    
}


plt.figure(figsize=(12,6))
sns.boxplot(x='model', y='perplexity', data=df, palette=custom_palette)
plt.title('Distribución de Perplexity por Modelo')
plt.yscale('log')
plt.tight_layout()
plt.show()

plt.figure(figsize=(15,8))
sns.barplot(x='prompt', y='perplexity', hue='model', data=df, palette=custom_palette)
plt.xticks(rotation=90)
plt.yscale('log')
plt.title('Perplexity por Prompt y Modelo')
plt.tight_layout()
plt.show()
```
