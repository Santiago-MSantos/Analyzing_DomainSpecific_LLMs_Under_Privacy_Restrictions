```python
import numpy as np
import plotly.graph_objects as go
from collections import defaultdict


def filtered_mean(data):
    filtered = [v for v in data if np.isfinite(v)]
    return np.mean(filtered)


raw_data = [
    ('DPSGD', 'Without context', 'Constant hints', dpsgd_constantHintsNoContext),
    ('SGD', 'Without context', 'Constant hints', sgd_constantHintsNoContext),
    ('DPSGD', 'With context', 'Constant hints', DPSGD_ConstantHints_WithContext),
    ('SGD', 'With context', 'Constant hints', SGD_ConstantHints_WithContext),
    ('SGD', 'With context', 'Variant hints', SGD_variableHints_withContext),
    ('SGD', 'Without context', 'Variant hints', sgd_variantHints_withoutContext),
    ('DPSGD', 'With context', 'Variant hints', dpsgd_variantHints_WithContext),
    ('DPSGD', 'Without context', 'Variant hints', dpsgd_variantHints_WithoutContext),
]


entries = [
    {
        "optimizer": opt,
        "context": ctx,
        "hint": hint,
        "avg": filtered_mean(values)
    }
    for opt, ctx, hint, values in raw_data
]


def get_grouped_trace(group_by):
    grouped = defaultdict(list)
    for e in entries:
        grouped[e[group_by]].append(e)

    sorted_keys = sorted(grouped.keys())
    if len(sorted_keys) != 2:
        raise ValueError("Solo se permiten comparaciones binarias")


    sort_criteria = lambda x: (x['hint'], x['context'] if x['context'] != sorted_keys[0] else "A"+x['context'], x['optimizer'])
    
    group1 = sorted(grouped[sorted_keys[0]], key=sort_criteria)
    group2 = sorted(grouped[sorted_keys[1]], key=sort_criteria)

    labels = []
    values = []
    text = []
    colors = []
    

    color_palette = [
        ('rgb(220,57,18)', 'rgba(220,57,18,0.6)'),   # Rojo y rojo transparente
        ('rgb(16,150,24)', 'rgba(16,150,24,0.6)'),   # Verde y verde transparente
        ('rgb(51,102,204)', 'rgba(51,102,204,0.6)'), # Azul y azul transparente
        ('rgb(153,0,153)', 'rgba(153,0,153,0.6)')    # Púrpura y púrpura transparente
    ]


    for i, (e1, e2) in enumerate(zip(group1, group2)):

        color_pair = color_palette[i % len(color_palette)]
        

        pair_id = f"Pair{i+1}"
        labels.append(f"[{e1['optimizer']}] [{e1['context']}] [{e1['hint']}]")
        values.append(e1['avg'])
        text.append(f"{e1['avg']:.2f}")
        colors.append(color_pair[0])


        labels.append(f"[{e2['optimizer']}] [{e2['context']}] [{e2['hint']}]")
        values.append(e2['avg'])
        text.append(f"{e2['avg']:.2f}")
        colors.append(color_pair[1])  


        if i < len(group1) - 1:
            labels.append(f"gap{i}") 
            values.append(None)  
            text.append("")
            colors.append('rgba(0,0,0,0)')  

    return labels, values, text, colors


def prepare_args(group_by):
    x, y, text, colors = get_grouped_trace(group_by)
    return [{"x": [x], "y": [y], "text": [text], "marker": {"color": colors}}]


initial_labels, initial_values, initial_text, initial_colors = get_grouped_trace('context')

fig = go.Figure(go.Bar(
    x=initial_labels,
    y=initial_values,
    text=initial_text,
    textposition="auto",
    marker_color=initial_colors
))


fig.update_layout(
    title="Exposure per Configuration. Using 5500 secrets with 5-15 repetitions",
    yaxis_title="Average Exposure",
    updatemenus=[ 
        dict(
            type="buttons",
            direction="right",
            showactive=True,
            buttons=[
                dict(label="Group by Optimizer", method="update", args=prepare_args("optimizer")),
                dict(label="Group by Context", method="update", args=prepare_args("context")),
                dict(label="Group by Hint Type", method="update", args=prepare_args("hint")),
            ],
            x=0.1, xanchor="left",
            y=1.15, yanchor="top"
        )
    ]
)

fig.update_layout(height=500)
fig.show()
```
