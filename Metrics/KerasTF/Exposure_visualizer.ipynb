```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


data = {
    "Repetition": [1000, 10000, 100000],
    "avg_exposure": [1.61, 2.12, 6.85],
    "std_dev": [1.11, 1.35, 0.93]
}


df = pd.DataFrame(data)


np.random.seed(42)
expanded_data = []
for i in range(len(df)):
    values = np.random.normal(df["avg_exposure"][i], df["std_dev"][i], 100)  
    for v in values:
        expanded_data.append({"Repetition": df["Repetition"][i], "avg_exposure": v})

df_expanded = pd.DataFrame(expanded_data)

plt.figure(figsize=(8, 5))


sns.boxplot(x="Repetition", y="avg_exposure", data=df_expanded, color="lightgreen", width=0.5)

plt.xlabel("Repetition")
plt.ylabel("Avg Exposure")
plt.title("Avg Exposure for different repetitions")
plt.grid(True)


plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np


repetitions = [1000, 10000, 100000]
avg_exposure = [1.61, 2.12, 6.85]
std_dev = [1.11, 1.35, 0.93]


plt.figure(figsize=(8, 5))


plt.plot(repetitions, avg_exposure, 'r:o', label="HS Canaries (No DP)")
plt.fill_between(repetitions, 
                 np.array(avg_exposure) - np.array(std_dev), 
                 np.array(avg_exposure) + np.array(std_dev), 
                 color='red', alpha=0.2)

plt.xscale('log') 
plt.xlabel("Count")
plt.ylabel("Exposure")
plt.title("Secret Exposure as a function of occurrences during training")
plt.legend()
plt.grid(True, which="both", linestyle="--", linewidth=0.5)


plt.ylim(0, 14)

plt.show()
```
