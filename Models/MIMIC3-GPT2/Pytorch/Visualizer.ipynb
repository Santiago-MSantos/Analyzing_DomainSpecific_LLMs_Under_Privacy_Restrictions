```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.colors import LinearSegmentedColormap, Normalize
import matplotlib.cm as cm




learning_rates = np.array([d["learningRate"] for d in data])
log_learning_rates = np.log10(learning_rates)
batch_sizes = np.array([d["batchSize"] for d in data])
warmup_steps = np.array([d["warmupSteps"] for d in data])
accuracies = np.array([d["accuracy"] for d in data])


colors = [
    (1, 0, 0),   # Red
    (1, 0.5, 0), # Orange
    (1, 1, 0),   # Yellow
    (0.5, 1, 0), # Yellow-green
    (0, 1, 0)    # Green
]
accuracy_cmap = LinearSegmentedColormap.from_list("accuracy_cmap", colors)


norm = Normalize(vmin=min(accuracies), vmax=max(accuracies))
color_values = [accuracy_cmap(norm(a)) for a in accuracies]

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')


scatter = ax.scatter(log_learning_rates, batch_sizes, warmup_steps, c=color_values, s=150, edgecolors='k')

# Add vertical lines from each dot to the floor plane (z=0)
for lr, bs, ws, col in zip(log_learning_rates, batch_sizes, warmup_steps, color_values):
    ax.plot([lr, lr], [bs, bs], [0, ws], color=col, linestyle='dashed', linewidth=1.5)


ax.set_xlabel("Learning Rate")
ax.set_ylabel("Batch Size")
ax.set_zlabel("Warmup Steps")
ax.set_title("Hyperparameter Tuning - 3D Accuracy Visualization")


ax.view_init(elev=20, azim=45)


sm = cm.ScalarMappable(cmap=accuracy_cmap, norm=norm)
cbar = plt.colorbar(sm, ax=ax, shrink=0.5)
cbar.set_label("Accuracy")

plt.show()
```

```python
# Re-attempting the visualization

import plotly.graph_objects as go
import numpy as np
from sklearn.linear_model import LinearRegression

# Filter data (removing low learning rates below 3e-06)
filtered_data = [d for d in data if d["learningRate"] >= 3e-06]

# Extract values for scatter plot
learning_rates = np.array([d["learningRate"] for d in filtered_data])
batch_sizes = np.array([d["batchSize"] for d in filtered_data])
warmup_steps = np.array([d["warmupSteps"] for d in filtered_data])
accuracies = np.array([d["accuracy"] for d in filtered_data])

# Convert learning rate to log scale for linear regression
log_learning_rates = np.log10(learning_rates)

# Prepare regression model
X = np.column_stack((log_learning_rates, batch_sizes, warmup_steps))
y = accuracies

# Fit linear regression
reg = LinearRegression().fit(X, y)

# Generate trend line (arrow direction)
start = np.array([log_learning_rates.min(), batch_sizes.min(), warmup_steps.min()])
end = np.array([log_learning_rates.max(), batch_sizes.max(), warmup_steps.max()])

# Predict accuracy for start and end points
accuracy_start = reg.predict([start])[0]
accuracy_end = reg.predict([end])[0]

# Convert back log learning rate to normal scale for visualization
start[0] = 10**start[0]
end[0] = 10**end[0]

# Create 3D scatter plot
fig = go.Figure()

# Scatter plot
fig.add_trace(go.Scatter3d(
    x=learning_rates, 
    y=batch_sizes,
    z=warmup_steps,
    mode='markers',
    marker=dict(
        size=8,
        color=accuracies,  # Color represents accuracy
        colorscale='Inferno',
        colorbar=dict(title="Accuracy"),
    ),
    text=[f"LR: {lr:.0e}, BS: {bs}, WS: {ws}, Acc: {acc:.3f}" 
          for lr, bs, ws, acc in zip(learning_rates, batch_sizes, warmup_steps, accuracies)],
    hoverinfo="text"
))

# Add trend line (arrow)
fig.add_trace(go.Scatter3d(
    x=[start[0], end[0]],
    y=[start[1], end[1]],
    z=[start[2], end[2]],
    mode='lines+markers',
    line=dict(color='cyan', width=5),
    marker=dict(size=5, color='cyan'),
    name="Trend Line"
))

# Fix axes
fig.update_layout(
    scene=dict(
        xaxis=dict(title="Learning Rate", type="log"),
        yaxis=dict(title="Batch Size"),
        zaxis=dict(title="Warmup Steps"),
    ),
    title="Hyperparameter exploration in SGD Finetuning"
)

fig.show()
```
