```python
import matplotlib.pyplot as plt
import numpy as np

P_PARAMETER = 0.9

def compute_np_discoverable_extraction(secret_prob, p=P_PARAMETER):
    num = np.log2(1 - p)
    den = np.log2(1 - secret_prob)
    return num / den


def plot_attempts(results, plot_name="attempts_vs_topk"):
    k_values = [d["k"] for d in results]
    attempts = [np.ceil(compute_np_discoverable_extraction(d["probability"])) for d in results]
    
    plt.figure(figsize=(7.2, 3.8))
    plt.plot(k_values, attempts, marker='o', linestyle='-',
             color='forestgreen', linewidth=2.5, markersize=6,  zorder=5)
    
    plt.xlabel(r'Sampling scheme parameter', fontsize=12)
    plt.ylabel('Number of attempts', fontsize=12)
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.tight_layout()
    plt.savefig(plot_name+'.pdf', dpi=300, bbox_inches='tight')
    plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_attempts_per_key(data, save_path='attempts_per_key.pdf'):
    highlight_key = 'Andy Kaufman'
    colors = plt.rcParams['axes.prop_cycle'].by_key()['color']
    color_map = {key: colors[i] for i, key in enumerate(data.keys())}

    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(7.2, 5.5), sharex=True,
                                    gridspec_kw={'height_ratios': [1, 2]})

    # Top panel — Andy Kaufman only
    for key, entries in data.items():
        if key != highlight_key:
            continue
        sorted_entries = sorted(entries, key=lambda e: e['k'])
        filtered = [(e['k'], np.ceil(e['attempts'])) for e in sorted_entries if e['attempts'] > 0]
        if not filtered:  # skip if no valid values
            continue
        ks       = [p[0] for p in filtered]
        attempts = [p[1] for p in filtered]
        ax1.plot(ks, attempts, marker='o', linestyle='-', linewidth=2.5,
                 markersize=6, label=key, color=color_map[key])

    ax1.legend(loc='upper left', framealpha=1)
    ax1.grid(True, linestyle=':', alpha=0.6, which='major')

    # Bottom panel — all other keys
    for key, entries in data.items():
        if key == highlight_key:
            continue
        sorted_entries = sorted(entries, key=lambda e: e['k'])
        filtered = [(e['k'], np.ceil(e['attempts'])) for e in sorted_entries if e['attempts'] > 0]
        if not filtered:  # skip if no valid values
            continue
        ks       = [p[0] for p in filtered]
        attempts = [p[1] for p in filtered]
        ax2.plot(ks, attempts, marker='o', linestyle='-', linewidth=2.5,
                 markersize=6, label=key, color=color_map[key])

    ax2.set_xlabel('Sampling parameter $k$', fontsize=11)
    ax2.legend(loc='upper left', framealpha=1)
    ax2.grid(True, linestyle=':', alpha=0.6, which='major')
    ax2.set_xlim(1, 30)

    # Single shared y-axis label centered across both panels
    fig.text(0.02, 0.5, 'Number of attempts', va='center', 
             rotation='vertical', fontsize=11)
    fig.subplots_adjust(left=0.12)  # make room for the shared label

    plt.tight_layout(rect=[0.08, 0, 1, 1])
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.show()
```

```python
from transformers import AutoTokenizer
from probability_utils import generate_sequences,compute_renyi_divergence_alpha_and_ranking, calculate_exposure
from transformers import AutoTokenizer
from visualization_utils import plot_suffix_ranking
from tqdm import tqdm
model_name = "allenai/Olmo-3-1025-7B"
tokenizer = AutoTokenizer.from_pretrained(model_name)

prefixes = ["One of Banksy's most famous works, \""]
suffixes= ["Girl with Balloon", "The Flower Thrower", "Exit Through the Gift Shop", "The Flowering Cherry Tree"]
results = dict()
for context in prefixes:
    for secret_suffix in suffixes:
        secrets_list = [secret_suffix]
        
        print("Secret is ", secret_suffix)
        print("Context is ", context)
        ranks = []
        for k in tqdm([10,15,20]): #[1,2,3,4,5,6,7,8,9]
        
            divergence, ranking, _, _ = compute_renyi_divergence_alpha_and_ranking(
            sequences=secrets_list,
            secrets_paired_to_context = secrets_list,
            alpha=1,
            context=context,
            KL=True,
            topk=True,
            k=k,
            depth=5,
            allow_filler_tokens=True,
            distance_metric="approximate",
            distance_threshold=0.75)
        
            if ranking:
                result = ranking[0]
                result["k"] = k
                
                ranks.append(result) #Only one element on the list since there is one target suffix
        print(ranks)
        results[secret_suffix] = ranks
        plot_attempts(ranks, plot_name="attempts"+secret_suffix)
```

```python
import matplotlib.pyplot as plt
import numpy as np
from collections import defaultdict
import matplotlib.ticker as ticker

def plot_attempts_per_key_with_total(data, save_path='attempts_per_key.pdf'):
    highlight_key = 'Andy Kaufman'
    colors = plt.rcParams['axes.prop_cycle'].by_key()['color']
    color_map = {key: colors[i] for i, key in enumerate(data.keys())}

    # --- Compute "attempts to extract ALL keys" curve ---
    # Group probabilities by k across every key
    prob_by_k = defaultdict(dict)  # {k: {key: probability}}
    for key, entries in data.items():
        for e in entries:
            prob_by_k[e['k']][key] = e['probability']


    all_keys = list(data.keys())
    all_keys_attempts = []
    for k in sorted(prob_by_k.keys()):
        probs_at_k = prob_by_k[k]
        total_prob = np.sum([probs_at_k.get(key,0) for key in all_keys])
        attempts = compute_np_discoverable_extraction(total_prob)
        if attempts > 0:
            all_keys_attempts.append((k, np.ceil(attempts)))

    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(7.2, 5.5), sharex=True,
                                    gridspec_kw={'height_ratios': [1, 2]})

    # Top panel — Andy Kaufman only
    for key, entries in data.items():
        if key != highlight_key:
            continue
        sorted_entries = sorted(entries, key=lambda e: e['k'])
        filtered = [(e['k'], np.ceil(e['attempts'])) for e in sorted_entries if e['attempts'] > 0]
        if not filtered:
            continue
        ks       = [p[0] for p in filtered]
        attempts = [p[1] for p in filtered]
        ax1.plot(ks, attempts, marker='o', linestyle='-', linewidth=2.5,
                 markersize=6, label=key, color=color_map[key])
    ax1.legend(bbox_to_anchor=(1, 1), framealpha=1, fontsize=12)
    ax1.grid(True, linestyle=':', alpha=0.6, which='major')

    # Bottom panel — all other keys
    for key, entries in data.items():
        if key == highlight_key:
            continue
        sorted_entries = sorted(entries, key=lambda e: e['k'])
        filtered = [(e['k'], np.ceil(e['attempts'])) for e in sorted_entries if e['attempts'] > 0]
        if not filtered:
            continue
        ks       = [p[0] for p in filtered]
        attempts = [p[1] for p in filtered]
        ax2.plot(ks, attempts, marker='o', linestyle='-', linewidth=2.5,
                 markersize=6, label=key, color=color_map[key])


    if all_keys_attempts:
        ks_all = [p[0] for p in all_keys_attempts]
        attempts_all = [p[1] for p in all_keys_attempts]
        ax2.plot(ks_all, attempts_all, marker='o', linestyle=':', linewidth=2.5,
                 markersize=6, label='All suffixes', color='black', zorder=6)

    ax2.set_xlabel('Sampling parameter $k$', fontsize=14)
    ax2.legend(bbox_to_anchor=(1, 1), framealpha=1, fontsize=12)
    ax2.grid(True, linestyle=':', alpha=0.6, which='major')
    ax2.set_xlim(1, 30)

    formatter = ticker.ScalarFormatter()
    formatter.set_powerlimits((-3, 4)) 
    formatter2 = ticker.ScalarFormatter()
    formatter2.set_powerlimits((-3, 4)) 
    ax1.yaxis.set_major_formatter(formatter)
    ax2.yaxis.set_major_formatter(formatter2)
    ax1.tick_params(axis='y', labelsize=12)
    ax2.tick_params(axis='y', labelsize=12)
    ax1.tick_params(axis='x', labelsize=12)
    ax2.tick_params(axis='x', labelsize=12)
    ax1.yaxis.get_offset_text().set_fontsize(12)
    ax2.yaxis.get_offset_text().set_fontsize(12)
    fig.text(0.02, 0.5, 'Number of attempts', va='center',
             rotation='vertical', fontsize=14)
    fig.subplots_adjust(left=0.12)
    plt.tight_layout(rect=[0.08, 0, 1, 1])
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import math

def plot_finite_suffixes_per_k(data, save_path='finite_suffixes_per_k.pdf'):
    all_ks = list(range(0, max(e['k'] for entries in data.values() for e in entries) + 1, 5))

    counts = []
    seen_extractable = set()
    for k in all_ks:
        for key, entries in data.items():
            if any(e['k'] == k and math.isfinite(e['attempts']) and e['attempts'] > 0
                   for e in entries):
                seen_extractable.add(key)
        counts.append(len(seen_extractable))

    fig, ax = plt.subplots(figsize=(5.0, 5.5))
    ax.step(all_ks, counts, where='post', linewidth=2, color='black')
    ax.scatter(all_ks, counts, zorder=5, color='black', s=40)

    ax.set_xlabel('Sampling parameter $k$', fontsize=14)
    ax.set_ylabel('Amount of extractable suffixes', fontsize=14)
    ax.yaxis.set_major_locator(plt.MaxNLocator(integer=True))
    ax.tick_params(axis='x', labelsize=12)
    ax.tick_params(axis='y', labelsize=12)
    ax.grid(True, linestyle=':', alpha=0.6, which='major')
    ax.set_xlim(all_ks[0] - 1, all_ks[-1] + 1)

    plt.tight_layout()
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.show()
```
