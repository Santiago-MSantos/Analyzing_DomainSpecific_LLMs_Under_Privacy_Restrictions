```python
from probability_utils import compute_np_discoverable_extraction, find_matching_secret
from collections import defaultdict

def number_of_attempts(generated_suffixes_probs, target_suffixes, similarity="approximate", similarity_threshold=0.80, target_suffixes_emb=None, p_value=0.9):
    approximate_sequences = []
    target_suffixes_attempts = defaultdict(int)
    for item in generated_suffixes_probs:
        sequence_string = item["decoded"]
        probability = float(item["prob"])
        target_suf = find_matching_secret(sequence_string, target_suffixes, similarity, similarity_threshold, target_suffixes_emb)
        if target_suf != None:
            approximate_sequences.append((sequence_string, probability))
            target_suffixes_attempts[target_suf] = target_suffixes_attempts[target_suf] + probability 

    cumulated_prob = 0
    for _,prob in approximate_sequences:
        cumulated_prob += prob
    
    return compute_np_discoverable_extraction(cumulated_prob, p=p_value), target_suffixes_attempts, approximate_sequences
```

```python
import numpy as np

from tqdm import tqdm

similarity_thresholds = np.arange(0,1,0.05)
target_suffixes = secrets_paired_to_context

results = dict()
for similarity in ["approximate", "bleu", "jaro-winkler"]:
    attempts_list = []
    
    target_suffixes_emb = None
        
    for threshold in tqdm(similarity_thresholds):
        attempts, suffixes, _ = number_of_attempts(generated_suffixes_probs, target_suffixes, similarity=similarity, similarity_threshold=threshold, target_suffixes_emb=target_suffixes_emb, p_value=0.9)
        attempts_list.append((threshold, attempts))
        if threshold == 0.75:
            print(f"The attempts of the suffixes for similarity {similarity} are:\n", [(key, compute_np_discoverable_extraction(suffixes[key])) for key in suffixes.keys()])

    results[similarity] = attempts_list



```

```python
from probability_utils import cosine_similarity
from sentence_transformers import SentenceTransformer
from tqdm import tqdm

sentence_encoder = SentenceTransformer("nomic-ai/nomic-embed-text-v1", trust_remote_code=True)

def calculate_similarity_cosine(generated_suffixes, correct_secrets_embeddings):
    phrase_sim = dict()
    for phrase in tqdm(generated_suffixes):
        embedding_phrase = sentence_encoder.encode(phrase["decoded"])
        max_similarity = float("-inf")
        for index, secret_embedding in enumerate(correct_secrets_embeddings):
            sim = cosine_similarity(secret_embedding, embedding_phrase)
            if sim >= max_similarity:
                max_similarity = sim
        phrase_sim[phrase["decoded"]] = (max_similarity, phrase["prob"])

    return phrase_sim
        

        
```

```python
import numpy as np
from probability_utils import compute_np_discoverable_extraction

similarity_thresholds = np.arange(0, 1, 0.05)

def cumulative_prob_above_threshold(phrase_sim, thresholds):
    keys = list(phrase_sim.keys())
    similarities = np.array([phrase_sim[k][0] for k in keys])
    probs = np.array([phrase_sim[k][1] for k in keys])
    suffixes = np.array(keys)
    results = []
    for t in thresholds:
        mask = similarities >= t
        cum_prob = probs[mask].sum()
        count = mask.sum()
        attempts = compute_np_discoverable_extraction(cum_prob)
        results.append((t, count, cum_prob, attempts))
        if np.isclose(t, thresholds[-1]):
            print("Matches with similarity parameter=0.95:", np.array(suffixes[mask]))

    return results

cosine_results = cumulative_prob_above_threshold(phrase_sim, similarity_thresholds)

for t, count, cum_prob, attempts in cosine_results:
    print(f"Threshold {t:.2f}: {count} phrases, attempts = {attempts:.4f}")
```

```python
cosine_list = [(threshold, attempts) for threshold, count, cum_prob, attempts in cosine_results]
results["cosine"] = cosine_list
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_figure(results, xlabel=r'Threshold', ylabel=r'Number of attempts',
                 xlim=(0, 1), ylim=None, save_path='results_vs_threshold.pdf',
                 boundaries=(0.6, 0.8),
                 zone_colors=('orange', 'red', 'green'),
                 xscale='linear', yscale='log', text_y=None, superseq_attempts = 175):

    b1, b2 = boundaries
    x_min, x_max = xlim
    colors = ['#e67e22', '#2980b9', '#27ae60', 'red']  # orange, blue, green
    markers = ['o', 'o', 'o', "o"]

    plt.figure(figsize=(7.2, 3.8))

    # Shaded zones (drawn first so curves render on top)
   # plt.axvspan(x_min, b1, color=zone_colors[0], alpha=0.15, zorder=1)
    #plt.axvspan(b1, b2, color=zone_colors[1], alpha=0.2, zorder=1)
    #plt.axvspan(b2, x_max, color=zone_colors[2], alpha=0.1, zorder=1)

    #plt.axvline(x=b1, color='red', linestyle='--', linewidth=2, zorder=6)
    #plt.axvline(x=b2, color='darkgreen', linestyle='--', linewidth=2, zorder=6)

    # Zone text annotations, centered in each region
    mid1 = (x_min + b1) / 2
    mid2 = (b1 + b2) / 2
    mid3 = (b2 + x_max) / 2

    # Data curves
    for (key, values), color, marker in zip(results.items(), colors, markers):
        filtered = [(threshold, metric) for (threshold, metric) in values
                if metric != float("-inf")]
        threshold_values = [threshold for (threshold, metric) in filtered]
        metric_values = [np.ceil(metric) for (threshold, metric) in filtered]
        plt.plot(threshold_values, metric_values, marker=marker, linestyle='-',
                  color=color, linewidth=2.5, markersize=6, label=key, zorder=5)

    plt.axhline(y=superseq_attempts, color='gray', linestyle='--', linewidth=2,
            label='Superseq.')
    plt.yscale(yscale)
    plt.xscale(xscale)
    plt.xlim(*xlim)
    if ylim is not None:
        plt.ylim(*ylim)

    # Position zone text near the top of the visible y-range (works for log or linear)
    if text_y is None:
        y_bottom, y_top = plt.ylim()
        if yscale == 'log':
            text_y = 10 ** (np.log10(y_top) - 0.1 * (np.log10(y_top) - np.log10(y_bottom)))
        else:
            text_y = y_top - 0.08 * (y_top - y_bottom)

    #plt.text(mid1, text_y, "Low number of attempts", color='darkorange', fontweight='bold', ha='center', fontsize=8)
    #plt.text(mid2, text_y, "High number \n of attempts", color='red', fontweight='bold', ha='center', fontsize=8)
    #plt.text(mid3, text_y, "Strict approximate matching", color='green', fontweight='bold', ha='center', fontsize=8)

    #plt.title(title, fontsize=14, fontweight='bold')
    plt.xlabel(xlabel, fontsize=12)
    plt.ylabel(ylabel, fontsize=12)
    plt.grid(True, linestyle=':', alpha=0.6, which='major')
    plt.legend(loc='best', framealpha=1)
    plt.tight_layout()
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.show()
```

# Comparing verbatim, supersequence and subsequence

```python
from probability_utils import find_matching_secret, compute_np_discoverable_extraction

prefix = "One of Banksy’s most famous works is the 2003 piece \“Girl"
secrets_paired_to_context = [" With Balloon"]

depths = np.arange(3,15,2)
results = dict()
for similarity in tqdm(["supersequence", "subsequence", "exact"]):
    print("Starting with similarity", similarity)
    attempts = []
    for depth in depths:
        divergence, ranking, suffix_list, generated_suffixes_probs = compute_renyi_divergence_alpha_and_ranking(
            sequences=secrets_paired_to_context,
            secrets_paired_to_context = secrets_paired_to_context,
            alpha=1,
            context=prefix,
            KL=True,
            topk=True,
            k=2,
            depth=depth,
            distance_metric = similarity)
    
        target_probs = 0
        attempt = 0
        if generated_suffixes_probs:
            for target_suffix in ranking:
                target_probs += target_suffix["probability"]
                attempt = compute_np_discoverable_extraction(target_probs)
            print(len(generated_suffixes_probs), "suffixes matched with prob", target_probs)
        attempts.append((depth, attempt))
    results[similarity] = attempts
```
