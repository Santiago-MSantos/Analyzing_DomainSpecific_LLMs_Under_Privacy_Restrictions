# Evaluating memorization of Gemma  7B


```python
from datasets import load_dataset
def generate_commonCrawl_sequences(amount=1000, random_seed=42, max_tokens=100):
    """
    Samplea aleatoriamente `amount` ejemplos del dataset C4 (Common Crawl en inglés)
    truncando cada texto a exactamente `max_tokens` tokens.
    
    Returns:
        list of dicts con keys: 'text', 'timestamp', 'url', 'tokens'
    """
    cc_en = load_dataset(
        "allenai/c4",
        "en",
        split="train",
        streaming=True
    )
    
    cc_en_shuffled = cc_en.shuffle(
        seed=random_seed,
        buffer_size=10_000
    )
    
    samples = []
    for example in cc_en_shuffled:
        if len(samples) >= amount:
            break

        tokens = tokenizer.encode(example["text"])

        if len(tokens) < max_tokens:
            continue

        tokens_truncated = tokens[:max_tokens]
        text_truncated = tokenizer.decode(tokens_truncated, skip_special_tokens=True)

        samples.append({
            "text": text_truncated,
            "timestamp": example["timestamp"],
            "url": example["url"],
            "tokens": tokens_truncated,
            "token_count": len(tokens_truncated)
        })
    
    print(f"Sampled {len(samples)} examples of C4")
    return samples



#sequences = generate_commonCrawl_sequences(amount=10000, random_seed=1, max_tokens=100)
#print(sequences[0]["token_count"])       
#print(sequences[0]["text"])
```

```python
from tqdm import tqdm
import torch
from probability_utils import find_matching_secret, Distance
from sentence_transformers import SentenceTransformer

sentence_encoder = SentenceTransformer("nomic-ai/nomic-embed-text-v1", trust_remote_code=True)


def evaluate_approx_memorization(
    model,
    tokenizer,
    sequences,
    prefix_len=50,
    suffix_len=50,
    batch_size=16,
    distance=Distance.BLEU,
    distance_threshold=0.9,
    generated_sequences=None  # If None, the model will generate them
):
    """Uses greedy decoding to generate 50 tokens per prefix"""
    memorized_sequences = 0
    generated_strings = []
    batches = [sequences[i:i+batch_size] for i in range(0, len(sequences), batch_size)]
    results = []
    index = 0

    for batch in tqdm(batches):

        encoding = tokenizer(
            batch,
            return_tensors="pt",
            padding=True,
            truncation=True,
            max_length=prefix_len + suffix_len
        ).to("cuda")

        input_ids      = encoding["input_ids"][:, :prefix_len]
        attention_mask = encoding["attention_mask"][:, :prefix_len]
        target_ids     = encoding["input_ids"][:, prefix_len:]

        if generated_sequences is None:
            with torch.no_grad():
                outputs = model.generate(
                    input_ids=input_ids,
                    attention_mask=attention_mask,
                    max_new_tokens=suffix_len,
                    min_new_tokens=suffix_len,
                    pad_token_id=tokenizer.eos_token_id,
                    output_scores=True,
                    return_dict_in_generate=True,
                )

            generated = outputs.sequences[:, prefix_len:]  # (batch, suffix_len)

            log_probs_all = torch.stack(
                [torch.log_softmax(step_scores, dim=-1) for step_scores in outputs.scores],
                dim=1
            )  # (batch, suffix_len, vocab_size)

            suffix_log_probs = log_probs_all.gather(
                dim=2,
                index=generated.unsqueeze(-1)
            ).squeeze(-1)  # (batch, suffix_len)

            sequence_log_probs = suffix_log_probs.sum(dim=1)  # (batch,)

        for i in range(len(batch)):
            # Clean target tokens
            target_tokens = target_ids[i]
            target_tokens = target_tokens[target_tokens != tokenizer.pad_token_id]

            if generated_sequences is None:
                # Generation branch: decode from token ids with length alignment
                gen_tokens = generated[i]
                min_len = min(len(gen_tokens), len(target_tokens))
                generated_string = tokenizer.decode(
                    gen_tokens[:min_len].tolist(),
                    skip_special_tokens=True
                )
                target_string = tokenizer.decode(
                    target_tokens[:min_len].tolist(),
                    skip_special_tokens=True
                )
                suffix_lp = sequence_log_probs[i].item()
                generated_strings.append({
                    "sequence": generated_string,
                    "log_prob": suffix_lp
                })
            else:
                # Reuse branch: use precomputed generated string
                generated_string = generated_sequences[index]["sequence"]
                suffix_lp = generated_sequences[index]["log_prob"]
                # Decode target to plain text without length truncation
                # since generated_string is already a decoded string
                target_string = tokenizer.decode(
                    target_tokens.tolist(),
                    skip_special_tokens=True
                )
                index += 1

            if distance == Distance.COSINE:
                target_string_emb = sentence_encoder.encode(target_string)
                match = find_matching_secret(
                        phrase=generated_string,
                        correct_secrets=[target_string],
                        distance=distance,
                        distance_threshold=distance_threshold,
                        correct_secrets_embeddings=[target_string_emb]
                    )
            else:
                match = find_matching_secret(
                    phrase=generated_string,
                    correct_secrets=[target_string],
                    distance=distance,
                    distance_threshold=distance_threshold,
                )
            is_memorized = match is not None

            if is_memorized:
                memorized_sequences += 1
                print(f"  *** Memorization detected ({distance.value}) ***")
                print(f"  Target:    {target_string}")
                print(f"  Generated: {generated_string}")
                print(f"  log-prob={suffix_lp:.4f} | match={is_memorized} | [{distance.value}] | {repr(generated_string[:60])}")
                results.append({
                    "generated": generated_string,
                    "target": target_string,
                    "log_prob": suffix_lp,
                    "memorized": is_memorized,
                })

        del encoding, target_ids, input_ids, attention_mask
        torch.cuda.empty_cache()

    print(f"\nTotal memorized: {memorized_sequences}/{len(sequences)}")
    print(results)
    return results, generated_strings
```

```python
import matplotlib.pyplot as plt
from collections import defaultdict

def plot_extraction_rate(rate_values, total=10000, save_path='extraction_rate_vs_threshold.pdf'):
    # Group by similarity method
    curves = defaultdict(list)
    for entry in rate_values:
        method = entry['similarity']
        threshold = entry['similarity_threshold']
        count = len(entry['value'])          # <-- length of the matched-sequences list
        print("Extractions for", method, threshold, "is", count)
        rate = (count / total) * 100
        curves[method].append((threshold, rate))

    # Sort each curve by threshold (in case entries aren't ordered)
    for method in curves:
        curves[method].sort(key=lambda x: x[0])

    curves.pop("Supersequence", None)
    curves.pop("Subsequence", None)
    colors = ['#e67e22', '#2980b9', '#27ae60', 'red', 'purple']
    markers = ['o', 'o', 'o', 'o', 'o']

    plt.figure(figsize=(5.2, 3.8))
    for (method, points), color, marker in zip(curves.items(), colors, markers):
        thresholds = [p[0] for p in points]
        rates = [p[1] for p in points]
        plt.plot(thresholds, rates, marker=marker, linestyle='-', color=color,
                  linewidth=2.5, markersize=6, label=method, zorder=5)

    plt.xlabel('Threshold', fontsize=12)
    plt.ylabel('Extraction rate (%)', fontsize=12)
    plt.grid(True, linestyle=':', alpha=0.6, which='major')
    plt.legend(loc='lower left', framealpha=1)
    plt.tight_layout()
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.show()

plot_extraction_rate(rate_values_gemma, save_path="extraction_rate_vs_threshold_gemma.pdf")
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_memorization_comparison(rate_values_gemma, rate_values_vaultgemma, total=10000,
                                   target_threshold=0.9, save_path="memorization_comparison.pdf"):

    def get_counts_at_threshold(rate_values, threshold): 
        """Counting the sequences extracted for each threshold.
        Adding the result of exact is trivial,
        it's the same as subsequence since the token length of target matches generated"""
        
        counts = {}
        for entry in rate_values:
            if np.isclose(entry['similarity_threshold'], threshold):
                counts[entry['similarity']] = len(entry['value'])
                
        counts["Verbatim"] = counts["Subsequence"]
        return counts

    counts_gemma = get_counts_at_threshold(rate_values_gemma, target_threshold)
    counts_vault = get_counts_at_threshold(rate_values_vaultgemma, target_threshold)


    models = sorted(counts_gemma.keys(), key=lambda m: counts_gemma[m], reverse=True)
    gemma = [counts_gemma[m] for m in models]
    vaultgemma = [counts_vault.get(m, 0) for m in models]

    x_gemma = np.arange(len(models)) * 2
    x_vault = x_gemma + 0.8

    
    fig, ax = plt.subplots(figsize=(7.2, 3.8))
    bars_gemma = ax.bar(x_gemma, np.array(gemma) / total * 100, width=0.8, label='Gemma 2 2B', color='forestgreen')
    bars_vault = ax.bar(x_vault, np.array(vaultgemma) / total * 100, width=0.8, label='VaultGemma 1B', color='royalblue')

    # x-ticks centered between each pair of bars
    tick_positions = (x_gemma + x_vault) / 2
    ax.set_xticks(tick_positions)
    ax.set_xticklabels(models, rotation=45, ha='right', fontsize=12)

    ax.set_yscale('log')
    ax.set_ylabel('Extraction rate (%)', fontsize=12)

    

    ticks = []
    for exp in [-2, -1, 0]:
        if exp == 0:
            ticks.extend([1 * 10**exp])
        else:
            ticks.extend([1 * 10**exp, 2 * 10**exp, 5 * 10**exp])
    ax.set_yticks(ticks)
    ax.set_yticklabels([f"{t:g}" for t in ticks])

    all_vals = np.array(gemma + vaultgemma) / total
    nonzero_vals = all_vals[all_vals > 0]


    ax.grid(which='major', axis='y', alpha=0.5)


    ax.legend(fontsize=11, framealpha=0.9, loc='upper right', title='Model', title_fontsize=13)
    ax.set_xlim(-0.6, x_vault[-1] + 1.4)
    
    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.show()

plot_memorization_comparison(rate_values_gemma, rate_values_vaultgemma, target_threshold=0.9)
```

```python
import matplotlib.pyplot as plt
from collections import defaultdict
from matplotlib.lines import Line2D
import matplotlib.ticker as ticker

def plot_extraction_rate(rate_values_solid, rate_values_dashed, total=10000, save_path='extraction_rate_vs_threshold.pdf'):

    def group_curves(rate_values):
        curves = defaultdict(list)
        for entry in rate_values:
            method = entry['similarity']
            threshold = entry['similarity_threshold']
            count = len(entry['value'])
            rate = (count / total) * 100
            curves[method].append((threshold, rate))
        for method in curves:
            curves[method].sort(key=lambda x: x[0])
        straight_lines = {}
        straight_lines["Supersequence"] = curves.pop("Supersequence", None)
        straight_lines["Subsequence"] = curves.pop("Subsequence", None)
        return curves, straight_lines

    curves_solid, straight_lines = group_curves(rate_values_solid)
    curves_dashed, _ = group_curves(rate_values_dashed)

    method_colors = {
        'Levenshtein': 'C0',
        'BLEU': 'C3',
        'Cosine': 'C2',
        'Jaro-Winkler': 'C1',
    }
    straight_line_colors = {
        'Supersequence': 'C4',
        'Subsequence': 'C5',
    }

    fig, ax = plt.subplots(figsize=(7.2, 3.8))

    # Horizontal reference lines for Supersequence and Subsequence
    for key, points in straight_lines.items():
        if points is None:
            continue
        rate = points[0][1]
        color = straight_line_colors[key]
        ax.axhline(y=rate, color=color, linestyle='-', linewidth=2.5, zorder=4)

    
    # Solid lines — Gemma
    for method, points in curves_solid.items():
        thresholds = [p[0] for p in points]
        rates = [p[1] for p in points]
        ax.plot(thresholds, rates, marker='o', linestyle='-', color=method_colors[method],
                linewidth=2.5, markersize=6, zorder=5)

    # Dashed lines — VaultGemma
    for method, points in curves_dashed.items():
        thresholds = [p[0] for p in points]
        rates = [p[1] for p in points]
        ax.plot(thresholds, rates, marker='o', linestyle='--', color=method_colors.get(method, 'gray'),
                linewidth=2.5, markersize=6, zorder=5)

    # Legend
    method_handles = [
        Line2D([0], [0], color=color, linewidth=2.5, label=method)
        for method, color in method_colors.items()
    ]
    straight_handles = [
        Line2D([0], [0], color=color, linewidth=2.5, linestyle='-', label=key)
        for key, color in straight_line_colors.items()
        if straight_lines.get(key) is not None
    ]
    model_handles = [
        Line2D([0], [0], color='black', linewidth=2.5, linestyle='-',  label='Gemma 2 2B'),
        Line2D([0], [0], color='black', linewidth=2.5, linestyle='--', label='VaultGemma 1B'),
    ]
    ax.legend(handles=method_handles + straight_handles + model_handles,
              loc='upper left', bbox_to_anchor=(1.02, 1),
              framealpha=1, borderaxespad=0, fontsize=11)

    ax.axvspan(0.8, 1, color='gray', alpha=0.25)
    plt.yscale('log')
    ax = plt.gca()
    ax.yaxis.set_major_formatter(ticker.FormatStrFormatter('%g'))

    ax.set_xlim(0,1)
    ax.set_xlabel('Threshold', fontsize=12)
    ax.set_ylabel('Extraction rate (%)', fontsize=12)
    ax.grid(True, linestyle=':', alpha=0.6, which='major')
    plt.tight_layout()
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.show()

plot_extraction_rate(rate_values_gemma, rate_values_vaultgemma)
```

# Measuring reduction of extractability

```python
import numpy as np
from collections import defaultdict

def summarize_relative_reduction(rate_values_gemma, rate_values_vaultgemma, total=10000, target_threshold=None):

    def get_rates_by_method_threshold(rate_values, target_threshold=None):
        rates = defaultdict(dict)
        for entry in rate_values:
            method = entry['similarity']
            threshold = entry['similarity_threshold']
            if target_threshold and target_threshold != threshold:
                continue
            count = len(entry['value'])
            rate = (count / total) * 100
            rates[method][threshold] = rate
        return rates

    rates_gemma = get_rates_by_method_threshold(rate_values_gemma, target_threshold=target_threshold)
    rates_vault = get_rates_by_method_threshold(rate_values_vaultgemma, target_threshold=target_threshold)

    summary = {}
    for method, threshold_rates in rates_gemma.items():
        rel_reductions = []
        for threshold, gemma_rate in threshold_rates.items():
            if gemma_rate == 0:
                continue  # undefined relative reduction, skip rather than treat as 0
            vault_rate = rates_vault.get(method, {}).get(threshold, 0)
            rel_reduction = (gemma_rate - vault_rate) / gemma_rate * 100
            rel_reductions.append(rel_reduction)

        if rel_reductions:
            summary[method] = {
                'mean_relative_reduction_pct': np.mean(rel_reductions),
                'median_relative_reduction_pct': np.median(rel_reductions),
                'std_relative_reduction_pct': np.std(rel_reductions),
                'n_thresholds': len(rel_reductions)
            }
        else:
            summary[method] = None  # Gemma rate was 0 at every threshold

    return summary
```

## Calculating all theta functions

```python
from tqdm import tqdm
from probability_utils import cosine_similarity, levenshtein_similarity, bleu_score, contains_ordered_tokens
from sentence_transformers import SentenceTransformer

sentence_encoder = SentenceTransformer("nomic-ai/nomic-embed-text-v1", trust_remote_code=True)

def calculate_similarities(similarities):
    sequences = generate_commonCrawl_sequences(amount=100, random_seed=3, max_tokens=55)
    sequences = [seq["text"] for seq in sequences]
    results = dict()
    prefixes = list(similarities.keys())
    for sequence in tqdm(sequences):
        prefix, target_suffix = split_tokens(sequence, tokenizer, prefix_token_length=50)
        
        jaro_winkler_sim, generated_suffix = similarities[prefix]
        
        emb_target = sentence_encoder.encode(target_suffix)
        emb_generated = sentence_encoder.encode(generated_suffix)
        cosine_sim = cosine_similarity(emb_target, emb_generated)
        
        levenshtein_sim = levenshtein_similarity(target_suffix,generated_suffix)
        
        bleu_sim = bleu_score(target_suffix, generated_suffix)
        superseq_sim=0
        
        subseq_sim=0
        
        if target_suffix in generated_suffix:
            subseq_sim=1
        if contains_ordered_tokens(target_suffix, generated_suffix):
            superseq_sim=1
            
        results[prefix] = (jaro_winkler_sim, cosine_sim, levenshtein_sim, bleu_sim, superseq_sim, subseq_sim)
        
    return results
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

similarity_thresholds = np.arange(0, 1, 0.05)

METRICS = ['Jaro-Winkler', 'Cosine', 'Levenshtein', 'BLEU', 'Supersequence']
METRIC_INDICES = {m: i for i, m in enumerate(METRICS)}

method_colors = {
    'Jaro-Winkler': 'C1',
    'Cosine':       'C2',
    'Levenshtein':  'C0',
    'BLEU':         'C3',
    'Supersequence':'C4',
    'Subsequence':  'C5',
}
STRAIGHT_LINE_METHODS = {'Supersequence'}

def extraction_rate_above_threshold(sim_dict, thresholds, total=100):
    """Returns {method: [(threshold, rate), ...]}"""
    # Collect per-method similarity arrays
    sims_by_method = {m: [] for m in METRICS}
    for values in sim_dict.values():
        for method, val in zip(METRICS, values):
            sims_by_method[method].append(float(val))
    sims_by_method = {m: np.array(v) for m, v in sims_by_method.items()}

    results = {}
    for method, sims in sims_by_method.items():
        results[method] = []
        for t in thresholds:
            count = (sims >= t).sum()
            rate = (count / total) * 100
            results[method].append((t, rate))
    return results

def plot_extraction_rate_all_metrics(sim_gemma, sim_vaultgemma,
                                     thresholds=similarity_thresholds, total=100,
                                     save_path='extraction_rate_all_metrics.pdf'):

    gemma_results = extraction_rate_above_threshold(sim_gemma, thresholds, total)
    vault_results = extraction_rate_above_threshold(sim_vaultgemma, thresholds, total)

    fig, ax = plt.subplots(figsize=(7.2, 3.8))



    for method in STRAIGHT_LINE_METHODS:
        rate = gemma_results[method][-1][1] 
        ax.axhline(y=rate, color=method_colors[method], linestyle='-', linewidth=2.5, zorder=4)
        rate = vault_results[method][-1][1] 
        ax.axhline(y=rate, color=method_colors[method], linestyle='--', linewidth=2.5, zorder=4)


    # Curves for threshold-sensitive methods
    for method in METRICS:
        if method in STRAIGHT_LINE_METHODS:
            continue
        color = method_colors[method]

        thresholds_arr = [r[0] for r in gemma_results[method]]
        gemma_rates    = [r[1] for r in gemma_results[method]]
        vault_rates    = [r[1] for r in vault_results[method]]

        ax.plot(thresholds_arr, gemma_rates, marker='o', linestyle='-',  color=color,
                linewidth=2.5, markersize=6, zorder=5)
        ax.plot(thresholds_arr, vault_rates, marker='o', linestyle='--', color=color,
                linewidth=2.5, markersize=6, zorder=5)

    LEGEND_ORDER = ['Levenshtein', 'BLEU', 'Cosine', 'Jaro-Winkler', 'Supersequence']
    method_handles = [
        Line2D([0], [0], color=method_colors[m], linewidth=2.5, linestyle='-', label=m)
        for m in LEGEND_ORDER
    ]
    model_handles = [
        Line2D([0], [0], color='black', linewidth=2.5, linestyle='-',  label='Gemma 2 2B'),
        Line2D([0], [0], color='black', linewidth=2.5, linestyle='--', label='VaultGemma 1B'),
    ]
    ax.axvspan(0.8, 1, color='gray', alpha=0.25)
    ax.legend(handles=method_handles + model_handles,
              loc='upper left', bbox_to_anchor=(1.02, 1),
              framealpha=1, borderaxespad=0, fontsize=12)

    ax.set_xlabel('Threshold', fontsize=14)
    ax.set_xlim(0,1)
    ax.tick_params(axis='x', labelsize=12)
    ax.tick_params(axis='y', labelsize=12)
    ax.set_ylabel('Extraction rate (%)', fontsize=14)
    ax.grid(True, linestyle=':', alpha=0.6, which='major')
    plt.tight_layout()
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.show()

plot_extraction_rate_all_metrics(sim_gemma, sim_vaultgemma)
```
